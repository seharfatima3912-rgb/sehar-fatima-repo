```python
# Import Path for using the existing raw and error folders.
from pathlib import Path

# Import datetime so failed files receive a unique timestamped name.
from datetime import datetime

# Point to the existing folders in the survey project root, one level above pipeline.
raw_folder = Path("..") / "raw"
error_folder = Path("..") / "error"

# Add your custom column names here in the same order as the Google Sheet columns.
# Keep this list empty until you are ready to provide the names.
CUSTOM_COLUMN_NAMES = ["timestamp", "python_skill", "career", "future_goal", "python_logics", "interests", "learning_style", "expectations", "pace", "difficulty", "balance"]

# Define the final CSV file and a temporary file used during writing.
raw_file = raw_folder / "google_sheet_data.csv"
temporary_file = raw_folder / "google_sheet_data.tmp.csv"

try:
    # Apply custom names only when you provide them.
    if CUSTOM_COLUMN_NAMES:
        # Check that the number of custom names matches the DataFrame columns.
        if len(CUSTOM_COLUMN_NAMES) != len(df.columns):
            raise ValueError(
                f"Provide {len(df.columns)} column names; "
                f"received {len(CUSTOM_COLUMN_NAMES)}."
            )

        # Replace the DataFrame column labels with your custom names.
        df.columns = CUSTOM_COLUMN_NAMES

    # Write the DataFrame without automatic column names or row numbers.
    df.to_csv(temporary_file, header=True, index=False)

    # Move the completed temporary file into its final raw-folder filename.
    temporary_file.replace(raw_file)

    # Confirm that the CSV was created successfully.
    print(f"CSV created successfully: {raw_file}")

except Exception as error:
    # Create a timestamp so each failed file has a unique name.
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    # Define the destination for the failed CSV in the existing error folder.
    failed_file = error_folder / f"google_sheet_data_failed_{timestamp}.csv"

    # Move the incomplete temporary file to error when it exists.
    if temporary_file.exists():
        temporary_file.replace(failed_file)
        print(f"Failed CSV moved to: {failed_file}")

    # Re-raise the original error so the notebook clearly reports the failure.
    raise error
```
