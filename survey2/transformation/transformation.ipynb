```python
# Import pandas so the CSV can be loaded if df is not already available.
import pandas as pd

# Import sqldf to run SQL queries against pandas DataFrames.
from pandasql import sqldf

# Load the existing raw CSV only when df is missing from the notebook session.
if "df" not in globals():
    df = pd.read_csv(r"D:\survey\raw\google_sheet_data.csv", skiprows=[1])

# Generate a sequential survey_key ordered by the CSV's timestamp column.
query = "SELECT ROW_NUMBER() OVER (ORDER BY timestamp) AS survey_key, * FROM df"

# Pass df explicitly so sqldf can register it as a SQL table named df.
result = sqldf(query, {"df": df})

# Display the resulting DataFrame.
result
```

```python
# Select the required columns from result, where survey_key was created.
query = "SELECT survey_key, python_skill, career, future_goal, expectations, pace, difficulty, balance FROM result"

# Pass result explicitly so sqldf can register it as a SQL table.
main_table = sqldf(query, {"result": result})

# Display the final transformed table.
main_table
```

```python
# Select the generated survey_key and the CSV's python_logics column from result.
# Alias python_logics as python_topics for this transformation output.
python_topics = sqldf(
    "SELECT survey_key, python_logics AS python_topics FROM result",
    {"result": result},
)

# Split each comma-separated answer into a Python list.
python_topics["python_topics"] = python_topics["python_topics"].str.split(",")

# Create one output row for each Python topic.
python_topics = python_topics.explode("python_topics", ignore_index=True)

# Display the transformed Python topics table.
display(python_topics)
```

```python
# Select the generated survey_key and interests column from result.
interests = sqldf(
    "SELECT survey_key, interests FROM result",
    {"result": result},
)

# Split each comma-separated response into a list of interests.
interests["interests"] = interests["interests"].str.split(",")

# Create one output row for each interest.
interests = interests.explode("interests", ignore_index=True)

# Display the transformed interests table.
display(interests)
```

```python
# Select the generated survey_key and learning_style column from result.
learning_style = sqldf(
    "SELECT survey_key, learning_style FROM result",
    {"result": result},
)

# Split each comma-separated response into a list of learning styles.
learning_style["learning_style"] = learning_style["learning_style"].str.split(",")

# Create one output row for each learning style.
learning_style = learning_style.explode("learning_style", ignore_index=True)

# Display the transformed learning styles table.
display(learning_style)
```

```python
# Import pandas for converting missing values to Python None values.
import pandas as pd

# Import PyArrow's table and Parquet-writing APIs directly.
import pyarrow as pa
import pyarrow.parquet as pq

# Convert pandas missing values to None so Arrow treats them as nulls.
normalized_table = main_table.astype(object).where(pd.notna(main_table), None)

# Build an Arrow table directly from Python column lists without pandas metadata.
arrow_table = pa.Table.from_pydict(
    {column: normalized_table[column].tolist() for column in normalized_table.columns}
)

# Write the Arrow table to the existing target path as Parquet.
pq.write_table(arrow_table, r"D:\survey\target\main_table.parquet")
```

```python
# Import pandas for converting missing values to Python None values.
import pandas as pd

# Import PyArrow's table and Parquet-writing APIs directly.
import pyarrow as pa
import pyarrow.parquet as pq

# Write a DataFrame without pandas extension metadata.
def write_parquet(dataframe, output_path):
    # Convert missing values to None so Arrow stores them as nulls.
    normalized = dataframe.astype(object).where(pd.notna(dataframe), None)

    # Build a plain Arrow table from the DataFrame's Python values.
    arrow_table = pa.Table.from_pydict(
        {column: normalized[column].tolist() for column in normalized.columns}
    )

    # Save the Arrow table as a Parquet file.
    pq.write_table(arrow_table, output_path)

# Save the learning styles table to the existing target folder.
write_parquet(learning_style, r"D:\survey\target\learning_style.parquet")

# Save the interests table to the existing target folder.
write_parquet(interests, r"D:\survey\target\interests.parquet")

# Save the Python topics table to the existing target folder.
write_parquet(python_topics, r"D:\survey\target\python_topics.parquet")
```
